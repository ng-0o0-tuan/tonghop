### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
n = 9
dem = 0
for k in range(1, n):
    if k > (n-1)/2:
        dem += 1
print(dem)
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
a, b, c = 4, 4, 4
p = (a+b+c)/2
A = p*(p-a)*(p-b)*(p-c)
s = math.sqrt(A)
print(s)
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
x = input("nhap chu cai ban dau: ")
n = int(input("khoang cach giua chu day va chu sau do la: "))
def dich_chuyen(x, n):
    if x.isupper(): #để kiểm tra chữ có viết hoa hay k
        moc = ord('A') #ord là mã ASCII của chữ in hoa (65-90)
    else:
        moc = ord('a') #ord là mã ASCII của chữ thường (97-122)
    vi_tri_cu = ord(x) - moc #để đưa về các số 1,2,3,4,...
    vi_tri_moi = (vi_tri_cu + n)%26 #đưa về khoảng cách số tiếp theo vd a=1 khoảng cách 1 thì chia phần dư cho 26 là 2 là vị trí chữ b
    return chr(vi_tri_moi + moc) #hàm ngược của hàm ord
print(dich_chuyen(x, n))
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
chu = str(input("nhap chu cai: "))
def chu_cai(chu):
    if len(chu) !=1: #len là để đếm số kí tự != 1 nghĩ là khác 1
        raise ValueError #dòng raisen là từ khóa để kích hoạt 1 lỗi khi gặp điều kiện k muốn
    else:
        return chu.lower()
print(chu_cai(chu))
```

### Problem 10
#### Write a function that, given a lowercase letter, returns a string containing that letter repeated 10 times.

```python
chu = str(input("nhap chu cai: "))
def lap_lai(chu):
    if len(chu) == 1 and chu.islower():
        return chu*10
print(lap_lai(chu)) #lưu ý nếu dùng biến thì k có ngoặc vd như trên còn nếu cùng chuỗi vd a thì cần "a",  chu ="c" "h" "u"
```
